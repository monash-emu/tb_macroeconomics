```python
import pandas as pd
from tb_macro.constants import DATA_PATH
```

```python
iso3 = "VNM"
all_data = pd.read_csv(DATA_PATH / "who/who_outcomes_20260514T0437Z.csv")
country_data = all_data[all_data["iso3"] == iso3]
```

```python
num_cols = [
    "new_sp_cur",
    "new_sp_cmplt",
    "new_snep_cmplt",
    "ret_cur",
    "ret_cmplt",
    "newrel_succ",
    "ret_nrel_succ",
    "mdr_succ",
]
denom_cols = [
    "new_sp_coh",
    "new_snep_coh",
    "ret_coh",
    "newrel_coh",
    "ret_nrel_coh",
    "mdr_coh",
]

num = country_data[num_cols].sum(axis=1)
denom = country_data[denom_cols].sum(axis=1)

tsr = pd.Series(num / denom)
tsr = tsr.where(denom > 0)
tsr.index = country_data["year"]
```
